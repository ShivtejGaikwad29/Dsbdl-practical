```python
import nltk
from nltk import pos_tag
from nltk.tokenize import TreebankWordTokenizer
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
document = "my name is shivtej sunil gaikwad. we can process the text with the natural language processing is better shivtej"
```

```python
tokenizer = TreebankWordTokenizer()
tokens = tokenizer.tokenize(document)
print(tokens)
```

```python
stp_words = set(stopwords.words('english'))
print(stp_words)
```

```python
pos = pos_tag(tokens)
print(pos)
```

```python
filtered = [word for word in tokens if word.lower() not in stp_words]
print(filtered)
```

```python
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered]
print(stemmed_tokens)
```

```python
lemma = WordNetLemmatizer()
lemma_tokens = [lemma.lemmatize(word) for word in filtered]
print(lemma_tokens)
```

```python
process_tokens = " ".join(lemma_tokens)
print(process_tokens)
```

```python
documents = [process_tokens]
```

```python
vector = TfidfVectorizer()
tfidf_mat = vector.fit_transform(documents)
name = vector.get_feature_names_out()
value = tfidf_mat.toarray()
for word,val in zip(name,value[0]):
    print(f"feature {word} : value {val:.4f}")
```
