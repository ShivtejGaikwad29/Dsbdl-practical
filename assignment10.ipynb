```python
df = sns.load_dataset('iris')
df
```

```python
df.hist(figsize=(10,8))
plt.suptitle("histogram for iris flower")
plt.tight_layout()
plt.show()
```

```python
for col in df.columns[:-1]:
    sns.boxplot(y=df[col],palette="Set2")
    # plt.tight_layout()
    plt.title(f"boxplot for {col}")
    plt.show()
```
